```python
def extract_questions(text):
    return re.findall(
        r'^\d+\.\s.*?(?=\n\d+\.|\Z)',
        text,
        re.MULTILINE | re.DOTALL
    )


def clean_question(q):
    replacements = {
        "⇥": "x",
        "": "<=",
        "≥": ">=",
        "−": "-",
        "→": "->"
    }

    for old, new in replacements.items():
        q = q.replace(old, new)

    lines = q.splitlines()
    lines = [line for line in lines if "page" not in line.lower()]

    return "\n".join(lines).strip()
```

```python
from pdf2image import convert_from_path
images = convert_from_path(path, dpi=300)
```

```python
from paddleocr import PaddleOCR
ocr = PaddleOCR(use_angle_cls=True, lang='en')
```

```python
import numpy as np

imp=np.array(images[2])
```

```python
result = ocr.ocr(imp)

# result: list of lines with (bbox, text)
lines = []
for line in result[0]:
    bbox, (txt, conf) = line
    x0, y0 = bbox[0]
    lines.append({"text": txt, "x": x0, "y": y0})
```

```python
lines
```

```python
lines = sorted(lines, key=lambda l: (round(l["y"]/10), l["x"]))
```

```python
def is_noise(t):
    t = t.lower()
    return any(k in t for k in ["important note", "page", "ms-"])

lines = [l for l in lines if not is_noise(l["text"])]
```

```python
import re

def is_q_start(t):
    return bool(re.match(r'^Q\.?\s*\d+', t.strip()))
```

```python
questions = []
current = []

for l in lines:
    if is_q_start(l["text"]) and current:
        questions.append(" ".join(current))
        current = []
    current.append(l["text"])

if current:
    questions.append(" ".join(current))
```
